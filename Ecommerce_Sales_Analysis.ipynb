```python
# Remove extra spaces from column names
df.columns = df.columns.str.strip()

# Check column names
df.columns
```

```python
# Calculate average values
df[["Age", "Quantity", "Sales", "Profit", "Ratings"]].mean()
```

```python
# Calculate total sales and profit
df[["Sales", "Profit"]].sum()
```

```python
# Calculate sales for each category
sales_by_category = df.groupby("Category")["Sales"].sum()

# Display the result
sales_by_category
```

```python
# Calculate profit for each category
profit_by_category = df.groupby("Category")["Profit"].sum()

# Display the result
profit_by_category
```

```python
# Calculate sales for each state
sales_by_state = df.groupby("State")["Sales"].sum()

# Sort from highest to lowest
sales_by_state.sort_values(ascending=False)
```

```python
# Calculate sales for each product
sales_by_product = df.groupby("Product Name")["Sales"].sum()

# Show highest sales first
sales_by_product.sort_values(ascending=False).head(10)
```

```python
# Count orders by status
orders_by_status = df["Order Status"].value_counts()

# Display the result
orders_by_status
```

```python
# Calculate sales by payment mode
sales_by_payment = df.groupby("Payment Mode")["Sales"].sum()

# Sort from highest to lowest
sales_by_payment.sort_values(ascending=False)
```

```python
# Calculate sales for each year
sales_by_year = df.groupby("Year")["Sales"].sum()

# Sort by year
sales_by_year.sort_index()
```

```python
# Calculate profit for each brand
profit_by_brand = df.groupby("Brand")["Profit"].sum()

# Sort from highest to lowest
profit_by_brand.sort_values(ascending=False)
```

```python
# Find highest and lowest sales
highest_sales = df["Sales"].max()
lowest_sales = df["Sales"].min()

# Display the results
print("Highest Sales:", highest_sales)
print("Lowest Sales:", lowest_sales)
```

```python
# Filter orders with positive profit
profitable_orders = df[df["Profit"] > 0]

# Show the first 5 profitable orders
profitable_orders.head()
```

```python
# Filter orders with negative profit
loss_orders = df[df["Profit"] < 0]

# Show the first 5 loss-making orders
loss_orders.head()
```

```python
# Count profitable and loss-making orders
print("Profitable Orders:", (df["Profit"] > 0).sum())
print("Loss-Making Orders:", (df["Profit"] < 0).sum())
```

```python
# Calculate average sales by category
avg_sales_category = df.groupby("Category")["Sales"].mean()

# Show highest average first
avg_sales_category.sort_values(ascending=False)
```

```python
# Find the product with the highest sales
best_product = df.groupby("Product Name")["Sales"].sum().idxmax()

# Display the product
print("Best-Selling Product:", best_product)
```

```python
# Count orders in each category
orders_by_category = df["Category"].value_counts()

# Display the result
orders_by_category
```

```python
# Calculate sales by shipping mode
sales_by_shipping = df.groupby("Shipping Mode")["Sales"].sum()

# Sort from highest to lowest
sales_by_shipping.sort_values(ascending=False)
```

```python
# Create a basic sales summary
summary = {
    "Total Sales": df["Sales"].sum(),
    "Total Profit": df["Profit"].sum(),
    "Average Sales": df["Sales"].mean(),
    "Average Profit": df["Profit"].mean()
}

# Display the summary
summary
```

```python
# Check dataset information
df.info()
```

```python
import matplotlib.pyplot as plt

# Create sales by category
sales_by_category = df.groupby("Category")["Sales"].sum()

# Create a bar chart
sales_by_category.plot(kind="bar")

# Add chart title and labels
plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

# Display the chart
plt.show()
```

```python
# Calculate sales by year
sales_by_year = df.groupby("Year")["Sales"].sum()

# Create a line chart
sales_by_year.plot(kind="line", marker="o")

# Add chart title and labels
plt.title("Sales Trend by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")

# Display the chart
plt.show()
```

```python
# Convert profit margin from text to numbers
df["Profit Margin %"] = pd.to_numeric(df["Profit Margin %"], errors="coerce")

# Calculate average profit margin
df["Profit Margin %"].mean()
```
